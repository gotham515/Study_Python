```python
# 사용자 정의 함수에 대헤서
#   - 어떻게 만들고
#   - 어떻게 사용하는지
#   - 사용시 주의점 (scope)

# Python이 기본적으로 가지고 있는 함수들
# Type() : 객체의 Class명을 return
# len()  : 길이를 return
# all()  : 집합형 자료구조에 대헤서 인자가 모두 참이면 true, 그렇지 않으면 false 를 return        =>  and
# any()  : 집합형 자료구조에 대해서 인자 중 하나라도 참이면 true, 그렇지 않으면 false 를 return   =>  or
# enumerate() : 순서가 있는 자료형(Sequence type)을 입력받아서 index를 포함하는 객체를 return     => 열거형
# map()  : 순서가 있는 자료형에 특정 함수를 각각 적용해서 map 객체를 return
#          반복 가능한 객체 (list, tuple, range)의 각 요소에 특정 함수를 적용할 수 있다. 그래서 그 결과를 담은 객체를 받아올 수 있다.

# 이런 함수들은 편하게 사용할 수 있는 함수이기 때문에
# 많이 알면 많이 알수록 좋다.

my_list = [1, 2, 3, 4, 5]
print(type(my_list))
print(len(my_list))

my_tuple = 1,
print(type(my_tuple))
print(len(my_tuple))

my_dict = { 'name' : '홍길동',
            'age' : 20,
            'address' : '서울'}
print(type(my_dict))
print(len(my_dict))

my_str = '이것은 소리없는 아우성!!'
print(type(my_str))
print(len(my_str))

my_list = [3.24, 100, '안녕', 0]
print(all(my_list)) # 모두 참일 경우만 True이다.
print(any(my_list)) # 하나라도 참이 있다면 True이다.

my_str = ['이것은', '소리없는', '아우성'] # index값과 각각 data를 쌍 튜플로 생성해주는데 새로 list 할당해야 한다.
result = enumerate(my_str)
print(result)                # <enumerate object at 0x7e8e57a4ca90>
my_enu = list(result)
print(my_enu)                # [(0, '이것은'), (1, '소리없는'), (2, '아우성')]
my_list = [tmp for idx, tmp in enumerate(['이것은 ', '소리없는 ', '아우성'])]
print(my_list)               # ['이것은 ', '소리없는 ', '아우성']
my_dict = { idx : tmp for idx, tmp in enumerate(['이것은 ', '소리없는 ', '아우성'])}
print(my_dict)               # {0: '이것은 ', 1: '소리없는 ', 2: '아우성'}

# map()
def my_square(x) :
    return x ** 2
my_number = [7, 8, 9, 10]
square_numbers = map(my_square, my_number)
print(list(square_numbers))
print(type(square_numbers))
```

```python
# 문제 1: 리스트의 합 구하기
# 주어진 리스트의 모든 요소의 합을 구하는 함수를 작성하세요.

def sum_of_list(lst):
    result = 0
    for tmp in lst :
        result += tmp
    return result

print(sum_of_list([1, 2, 3, 4]))  # 출력: 10
```

```python
# 문제 2: 리스트에서 최댓값과 최솟값 구하기
# 리스트에서 최댓값과 최솟값을 구하는 함수를 작성하세요.

def max_and_min(lst):
    max, min = 0, 0
    for tmp in lst :
        if max < tmp :
            max = tmp
        elif max > tmp :
            min = tmp
    return max, min

print(max_and_min([10, 20, 4, 45, 99]))  # 출력: (99, 4)
```

```python
# 문제 3: 리스트 뒤집기
# 주어진 리스트를 역순으로 뒤집는 함수를 작성하세요.
# (직접 구현해 보세요, list.reverse() 사용하지 말 것)

def reverse_list(lst):
    rev_list = []
    for idx in range(len(lst)-1, -1, -1) :
        rev_list.append(lst[idx])
    return rev_list

print(reverse_list([1, 2, 3, 4]))  # 출력: [4, 3, 2, 1]
```

```python
# 문제 4: 리스트 중복 제거
# 리스트에서 중복된 요소를 제거하고, 중복이 제거된 리스트를 반환하는 함수를 작성하세요.

def remove_duplicates(lst):
    set_list = set(lst)
    return list(set_list)

print(remove_duplicates([1, 2, 2, 3, 4, 4, 5]))  # 출력: [1, 2, 3, 4, 5]
```

```python
#문제 5: 리스트 요소들의 곱 구하기
# 리스트의 모든 요소의 곱을 구하는 함수를 작성하세요.

def multiply_list(lst):
    result = 1
    for tmp in lst :
        result *= tmp
    return result

print(multiply_list([1, 2, 3, 4]))  # 출력: 24
```

```python
# 문제 6: 리스트에서 짝수와 홀수 구분하기
# 주어진 리스트에서 짝수와 홀수를 구분하여 별도의 리스트에 저장한 뒤 반환하는 함수를 작성하세요.

def separate_even_odd(lst):
    lst_even = []
    lst_odd = []

    for tmp in lst :
        if tmp % 2 == 0 :
            lst_odd.append(tmp)
        else :
            lst_even.append(tmp)

    return lst_odd, lst_even

print(separate_even_odd([1, 2, 3, 4, 5, 6, 7, 8, 9]))  # 출력: ([2, 4, 6, 8], [1, 3, 5, 7, 9])
```

```python
# 문제 7: 리스트 요소들의 위치 교환
# 리스트의 첫 번째 요소와 마지막 요소의 위치를 바꾸는 함수를 작성하세요.

def swap_first_last(lst):
    tmp_index = lst[-1]
    lst[-1] = lst[0]
    lst[0] = tmp_index
    return lst

print(swap_first_last([1, 2, 3, 4, 5]))  # 출력: [5, 2, 3, 4, 1]
```

```python
# 문제 3: 키 값 뒤집기
# 주어진 딕셔너리에서 키와 값을 뒤집어 새로운 딕셔너리를 반환하는
# 함수를 작성하세요. (값은 고유하다고 가정합니다.)

def invert_dict(d):
    new_dict = {}
    for key in d :
        new_dict_key_value = d[key]
        new_dict[new_dict_key_value] = key
    return new_dict

print(invert_dict({'a': 1, 'b': 2, 'c': 3}))  # 출력: {1: 'a', 2: 'b', 3: 'c'}
```
