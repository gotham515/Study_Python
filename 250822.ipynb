```python
# class 내부의 method에 대해서
# 우리가 알고 있는 건 instance method 에요


class MyClass(object) :

    class_variable = 100 # class variable

    def __init__(self, name) :  # magic function : 객체 초기화 담당 => initalizer 라고 한다.
        self.name = name        # 생성자, 생성자 함수, constructor
                                # 일반적으로 property(속성)을 생성.......

    # 로직처리를 하기 위한 함수 => method
    # 이런 method를 instance method 라고 부른다.
    # 이 method를 이용하려면 반드시 instance(객체)를 생성해야 한다
    # method의 첫번재 인자로 self가 오면 이 method는 instance method이다.
    def my_method(self) :
        print('열심히 일을 해요!')

    # method 중 instance method가 아닌 다른 method도 있다.
    # class method가 있다.
    # 대신 class method 는 어떤 용도로 사요되는냐 -> 일반적으로 class variable를 제어 하기 위해
    # class method는 특수한 decorator를 이용한다.
    # decorator 는 @ 으로 표현

    @classmethod
    def change_class_variable(cls, num) :
        cls.class_variable = num # class variable을 편하게 사용할 수 있다.
```

```python
# method의 종류?
# 1. instance method
# 2. class method
# 3. static method
# 그럼 magic function(method)는 머지? 인스턴스 메소드인가??


class MyClass :

    price = 1000 # class variable // 구현이 좋다. 유지보수가 힘듬..

    @staticmethod # staticmethod 에 대한 decorator
    def my_sum(a, b) :
        return a + b

    @classmethod # classmethod에 대한 decorator
    def print_class_variable(cls, num) : # class method
        cls.price = num

    def print_info(self) :  # instance method
        print('안녕하세요! 반가워요!')


obj1 = MyClass()
obj1.print_info() # print_info()를 사용하려면 instance가 있어야 한다.

MyClass.print_class_variable(10)
# 일반적으로 class method는 class 명을 직접 이용해서 사용
# instance를 이용해서 사용하는 것도 가능 / name space 가 상위 이기 때문!

obj1.my_sum(10,20)
```

```python
# Generator
def my_func(data) :
    # empty list (비어있는 리스트)를 생성
    #result = []
    result = list()

    # list에 새로운 데이터를 새롭게 추가하려면 많이 사용하는 method -> result.append()
    for tmp in data :
        result.append(tmp*2) # 받아온 data의 값에 2를 곱하여 추가하겠다.
    return result


twice_list = my_func([1,2,3,4,5])
print(twice_list)
```

```python
# 위의 코드를 generator 라는걸 이용해서 살짝 수정해보자
# 머신러닝에서 이 generator 를 이용해서 데이터 처리를 합니다.

def my_func(data) : # 입력인자parameter로 list를 받아요

    for tmp in data :
        yield tmp * 2

twice_list = my_func([1,2,3])
print(twice_list) # << generator object
print(f'반복처리의 첫번째 값은 : {next(twice_list)}') # 2
print(f'반복처리의 두번째 값은 : {next(twice_list)}') # 4
print(f'반복처리의 세번째 값은 : {next(twice_list)}') # 6
```
