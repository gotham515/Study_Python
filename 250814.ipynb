```python
# 1. Numeric type(숫자)
# int(정수) - class
# float(실수) - class
# complex(복소수) - class

a = 100
print(a)
print(type(a)) #type이라는 명칭은 어떤 클래스로부터 파생되었는지 type으로 확인..??

b = 3.14
print(b)
print(type(b))

# 정리 : Python의 모든 값은 객체이다. 객체는 무조건 class로부터 만들어진다. (파생된다)
# 간단한 연산 예제
div = 3 / 4
print(div) # 결과는 0.75

result = 3 ** 3 # 지수연산 ( * 을 늘린다.)
print(result)

result = 10 % 3 # 나머지 연산
print(result)

result = 10 // 3 # 몫 연산
print(result)

# 프로그래밍에서 연산은 같은 데이터 타입끼리 연산이 수행된다.
# 그 결과도 같은 데이터 타입으로 만들어진다.
# 파이썬은 기본 연산 할 때 모든 숫자가 실수로 변환되서 연산된다..!
```

```python
# 2. Sequence (순차적인, 순서)
#   1) list : Python에서 가장 기본적인, 대표적인, 중요한 자료구조이면서 데이터 타입이다.
#             list는 class명이다.
#             임의의 객체를 순서대로 저장하는 집합형 자료구조. 다른언어의 배열과 유사
#             단, Python에는 배열이 없고 대신 list 사용
#             literal은 []으로 표현. (literal : 어떠한 개념을 사용하기위한 지정한 것)
#             용법은 이 외에도 상당히 많은 함수들이 있다.

a = [] # empty list 내용이 없는 list
a = [1,2,3,4,5] # list 사용
a = [1, 2.0, 3.14, 'hello', [4,5]]  #안에 객체만 들어가면 사용가능 list 중첩도 가능
a = [1, 2, ['show', 'me', 'the', 'money'], 3.14]
print(a)


#list를 사용하기 위해서 indexing과 slicing에 대해 이해해야 함
#   indexing : 순서(특정위치에 대한)를 이용해서 값을 알아내거나 값을 변경하는 방법
#   slicing  : list의 특정 부분만 알아내거나 변경할 때 사용
#   indexing과 slicing을 할 때는 [] 이용한다.

print(a[1]) # 2 (list는 0부터 시작하고 []번째의 값을 출력해준다.)
print(a[2][0])
print(a[-1])  # -indexing은 뒤에서부터 순차적으로 되고 파이썬에서만 제공해준다.
print(a[1:3]) # 첫번째 값은 포함하고 마지막값은 미포함하여 slicing한다. 원본과 같은 데이터 타입이 나온다.
print(a[:2])  # 첫번째 값 생략 가능 ( 맨 처음부터 시작할 경우)
print(a[1:])  # 두번째 값 생략 가능 ( 끝까지 확인 할 경우)
print(a[:])   # 값이 모두 생략 가능 ( 처음부터 끝까지 모두 slicing)
```

```python
#이어서
# list의 연산

a = [1,2,3]
b = [4,5,6]
print(a+b) # list + 연산 : 두 list를 연결해주는 의미이다.
print(a*3) # list * 연산 : 값 만큼 더하라는 의미 (연결)


# 파이썬에서 나오는 모든 값은 객체!!
# 객체는 반드시 class로부터 파생(만들어지는)
# 객체는 도대체 어떤거야!? >> 값과 기능(함수)가 같이 결합되어 있는 자료구조

a.append(4) # append는 인자값을 받아 a의 마지막에 추가해주는 기능(함수)
print(a)


# 예제
mylist = [2, 4, 6, 1, 8, 10, 5, 3]
#mylist.sort() # sort : 오름차순 정렬 기능(함수)
#mylist.reverse() # reverse : list의 순서를 역순으로 바꿔주는 기능(함수)
print(mylist)


# 객체가 가지는 기능(함수 또는 메소드)는 두가지 형태가 있다.
#  1. 객체가 가지는 데이터를 직접 변화시키는 함수 > sort, reverse 등
#  2. 객체가 가지는 데이터를 변화시키지 않지만 처리된 결과를 리턴해준다. ex) result = mylist.sort() 와 같이 결과를 객체로 새로 받아와야 함
```

```python
# 2. Sequence (순차적인, 순서)
#     2) tuple : list와 유사한 순서가 있는 집합형 자료구조
#              : list와 다르게 interal을 () 사용한다.

a = () # 비어 있는 tuple
print(type(a)) #class = tuple

# 요소가 1개인 tuple : 연산자 우선순위를 나타내어 ()에 있는 값의 class를 먼저 인식
a = (1)
print(type(a)) #class = int

#요소가 1개인 tuple을 사용할때는 , 사용해야한다!!
a = (1,)
print(type(a)) #class = tuple


a = (1,2,3,4) #일반적인 형태의 tuple
a = 1,2,3,4,  #tuple은 ()를 생략할 수 있다. 위와 동일하고 거의 생략해서 사용하고 있다.

# 결과적으로 파이썬 코드를 보면 아래와 같은 코드가 많이 보인다.
a,b,c = 10,20,30 # 생략된 코드 (a,b,c) = (10,20,30) >> a = 10, b = 20, c = 30


# 그럼 list와 tuple의 차이점은 무엇인가?? tuple은 read only으로 사용된다.


a = [1,2,3,4] # list
a[0] = 100 #indexing으로 값을 변경 > 0번째 값을 100으로 변경한다.
print(a)

a = 1,2,3,4 # ()생략한 tuple
#a[0] = 100  # tuple은 값을 변경할 수 없다!! error : 'tuple' object does not support item assignment
print(a)
```

```python
# 이어서
# list와 tuple은 class이다.
a = []      # 비어 있는 list
a = list()  # 비어 있는 list
a = ()      # 비어 있는 tuple
a = tuple() # 비어 있는 tuple

# 일반적으로 다음과 같이 사용한다
a = [1,2,3,4]
b = tuple(a) # list a를 tuple 로 변경하고 싶을 때
print(a) # [1, 2, 3, 4] > list
print(b) # (1, 2, 3, 4) > tuple

c = (1,2,3,4)
d = list(c)
print(c)
print(d)
```

```python
# 2. Sequence (순차적인, 순서)
#     3) range : 범위를 나타내는 의미
#                조금 특이한 데이터 타입
#                list와 tuple은 실제 데이터를 메모리에 가지고 있다.
#                값의 범위를 가지고 있다 라는 의미만 메모리에 저장되게 된다.

a = [1,2,3,4,5,6,7]
range1 = range(10,100) # 값의 범위를 가지고 있는 객체!! ( 두번째 값은 미포함되며 결과는 1 ~ 99 이다.)
print(range1)  # range(10, 100) 값을 가지고 있는것이 아니고 범위를 가지고 있는 것이다..

#그러면 indexing이 되나요? >> 가능
print(range1[2]) # 12

#그러면 slicing이 되나요?  >> 가능 (*원본 그대로 데이터타입 가지고 가기 때문에 range에서 slicing하면 결과도 range가 된다.)
print(range1[2:10]) #range(12, 20)
```

```python
# 3. Text Sequence type
#   파이썬의 문자열은 문자와 문자열의 구분이 없다.
#   파이썬은 모두 다 문자열이다. 따라서 '',""의 차이가 없다! >> 관용적으로 ''을 더 많이 사용한다.
#   일반적으로 다른 언어에서의 문자는 1글자('')를 지칭하고, 문자열은 여러글자("")를 지칭한다.

a = "안녕하세요!"
b = '하이루하이루'
c = '''이것은
소리 없는
아우성''' # 여러줄의 문자열을 생성해야 할 때는 '''...'''을 사용하여 표현해준다.
print(c)

# 문자열은 list안에 각각에 문자들이 들어가 있는 형태이다.
first = '이것은 '
second = '소리 없는 '
third = '아우성'
print(first + second + third)
```

```python
# 이어서
# *주의사항*
first = '이것은   '     # class : str
second = '소리 없는 ' # class : str
third = 100           # class : int
#print(first + second + third)  # 에러 str과 int list 연결할 수 없다.

print(first + second + str(third)) # int를 str변환 (str를 int로 변환할수있따업따?????????...)

myStr = 'This is a sample Text' # 문자열 str - 내부적으로 list로 구성되어 있기 때문에 indexing과 slicing이 가능하다!!

print(myStr[3]) # s (indexing)
print(myStr[3:10]) # s is a (slicing)
```

```python
from os import name
# 4. Mapping type
#       dict (class)를 이용해서 만든 객체를 파이썬에서는 dictionary라고 부른다.
#       key와 value를 쌍으로 데이터를 표현
#       literal은 {}을 사용한다.

a = {} # 비어있는 dictionary
print(type(a)) # class = dict

a = {'name' : '홍길동'}  # name 이 key, 홍길동이 value 로 쓰인다.
# 그러다 보니 value에 해당하는 값은 python에서 나오는 모든 값을 사용할 수 있다.
# 하지만 key는 변경이 안되는 값을 사용해야만 한다
# key 값으로 list가 올 수 없다 (list는 값이 변할 수 있기 때문에) tuple은 read only 이므로 사용가능하다.

#만약 데이터를 여러개 저장하려면
student = {
    'name' : '홍길동',
    'dept' : '컴퓨터',
    'score' : 3.5
    }

# indexing이 가능하다. 단 key 값을 이용해서 사용해야 한다.
print(student['score'])

# 아래와 같은 경우라면??
student = {
    'name' : '홍길동',
    'name' : '신사임당', #중복된 key 값이 있는 경우는? == 문법상 문제는 없지만 대신 key의 value값을 알수 없게 된다.
    'dept' : '컴퓨터',
    'score' : 3.5
    }# 이렇게 사용하면 안된다!!

student = {
    'name' : '홍길동',
    'dept' : '컴퓨터',
    'score' : 3.5,
    ('address',) : '서울'
}
student[('address',)] = '부산' #tuple 키의 value를 변경
print(student)
```

```python
# 이어서
# dictionary의 대표적 함수
#   1. keys()   : dict의 key들만
#   2. values() : dict의 value들만

student = {
    'name' : '홍길동',
    'dept' : '컴퓨터',
    'score' : 3.5
}

# keys()
print(student.keys()) # dict_keys(['name', 'dept', 'score']) key들을 list처럼 list는 아닌
print(type(student.keys()))

# values()
print(student.values()) # list처럼 하지만 list는 아닌
```
