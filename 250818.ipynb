```python
# Set Type
# ** Set은 데이터를 저장할 수 있는 주머니 또는 바구니
# ** 순서가 없다. Key도 없다 (dict와 달리) => 중복을 배제할 수 있다.! ( 같은 값이 있어도 하나만 작용)
# literal 은 어떤걸 사용?? => {}
# 대신 Set은 키와 값의 쌍이 아니라 단순한 값만 포함한다.

a1 = {'name' : '홍길동'}  # 이건 dict
a = {1, 2, 3}             # 이건 Set
print(type(a))

# Set Class의 Instance(객체)를 생성하세요 라는 의미이다.
a = set()
a = set([1, 2, 3, 4, 5])
print(a)

# set에 넣으면 list여도 중복을 배제시켜준다.
a = set([1, 2, 3, 4, 5, 4, 2, 1])
print(a)

# set은 수학적으로 집합을 의미하며 집합 연산을 제공한다 : 합집합, 교집합, 차집합
s1= set([1, 2, 3, 4, 5])
s2 = set([3, 4, 5, 6, 7])

# 합집합 두 가지 방법
print(s1 | s2)      # 연산자 이용
print(s1.union(s2)) # 함수 이용

# 차집합 두 가지 방법
print(s1 - s2)            # 연산자 이용
print(s1.difference(s2))  # 함수 이용

# 교집합 두 가지 방법
print(s1 & s2)              # 연산자 이용
print(s1.intersection(s2))  # 함수 이용

# set은 순서나 key를 가지고 있지 않아 indexing을 할 수 없다
a = set([1, 2, 3, 4, 5])

# set에 값 추가
a.add(10)
print(a)

# set에 값 삭제
a.remove(3)
print(a)
```

```python
# Built-In Type
# 6. Bool Type ( 논리 객체 )
# 사용하는 class는 bool
# 사용 할 수 있는 값은 단 2개 : True, False (대소문자 구분함)

# 주의 사항 ( 논리연산 시 )
# 1. 빈 문자열 => '', "" : False로 간주
# 2. 빈 list => [] : False로 간주
# 3. 빈 tuple => () : False로 간주
# 4. 빈 dict => {} : False로 간주
# 5. 숫자 0 : False로 간주 (숫자 0 이외의 다른 숫자는 True로 간주)

a = True
b = False
print(int(a)) # => 1
print(int(b)) # => 0
```

```python
# 날짜 타입에 대해

# Built-In Type이 아니다.
# 날짜를 사용하려면 당연히 import를 해야 한다. ( 외부 모듈을 사용하기 위한 추가)

# from 패키지 import 패키지 안에 존재하는 모듈
from datetime import date, datetime

today = date.today() # date 객체이므로 함수(기능)를 가지고 있다
print(today) # 해당 코드가 실행되고 있는 컴퓨터의 날짜를 가지고 온다.

today = datetime.today()# 타임존..??
print(today)

# 날짜에 대한 연산이 생각보다 쉽지 않다.
# 날짜 연산을 쉽게 하기 위해서 모듈을 제공
from datetime import date, datetime, timedelta

today = date.today()
days = timedelta(days = -20) # 날짜의 개념 가지고 있는 또는 지칭하는 객체이므로 연산을 쉽게 할 수 있다.
print(today + days)          # timedelta는 월과 연도는 사용할 수 없다. 일만 사용할 수 있다.
```

```python
from dateutil.relativedelta import relativedelta # colab에서 미리 제공을 해준 패키지?
today = date.today()
days = relativedelta(months = -2) # 월과 연도 사용하기 위해선 relativedelta 사용
print(today + days)
```

```python
# Control Statement(제어문)
# 조건에 따른 분기문 - if

# 특정 블록이 아닌데 들여쓰기하면 오류난다. 임의의 들여쓰기는 할 수 없다.

x = 5

if x > 5 :
  # 여기서부터 Bolck 시작 ( 들여쓰기가 자동으로 들어가게 됨 )
  print('x 값이 5보다 크다.')
elif x == 5 :
  print('x 값이 5와 같다.')
else :
  print('이도 저도 아니다.')


fruits = ['apple', 'banana', 'cherry'] # list 안에 문자열이 들어가있다.
if 'apple' in fruits : # 집합형 in(안에) 조건이 있다면!
  print('apple이 목록에 있어요')
  print('사과는 맛있다.')
else :
  print('apple이 목록에 없어요')
  print('사과를 먹고 싶어요')

# 집합형 자료구조 : list, tuple, range, str, dict, set
student_score = {'홍길동' : 100 , '이순신' : 90}
if '이순신' in student_score : # dict로 집합형이 올 경우 key 값으로 구분한다.
  print('이순신이 dict 안에 있다')
```

```python
# 반복할 횟수가 정해져 있을 때 - for
# 일반적으로 2가지 형태로 사용
# for ~ in range() -> 반복 횟수를 고정해서 사용할 때 10번, 100번
# for ~ in list, dict -> 집합형 자료구조 안에 있는 요소수 만큼 반복

# 1부터 100까지 숫자의 합을 구하기
my_sum = 0 # 값을 누적할 변수
my_range = range(1,101,1) # 1부터 100까지 1씩 증가하는 값들의 모임 *range(초기숫자(포함), 마지막숫자(미포함), 증가값)*

for tmp in my_range : # 100번 반복
  my_sum += tmp
print(my_sum)

a = [1,2,3,4,5]

for tmp in a :
  print(tmp, end = ' ')
```
