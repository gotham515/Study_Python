```python
# Built-In Type
# 5. Set Type
# ** Set은 데이터를 저장할 수 있는 주머니 또는 바구니
# ** 순서가 없다. Key도 없다 (dict와 달리) => 중복을 배제할 수 있다.! ( 같은 값이 있어도 하나만 작용)
# literal 은 어떤걸 사용?? => {}
# 대신 Set은 키와 값의 쌍이 아니라 단순한 값만 포함한다.

a1 = {'name' : '홍길동'}  # 이건 dict
a2 = {1, 2, 3}             # 이건 Set
print(type(a1))
print(type(a2))

# Set Class의 Instance(객체)를 생성하세요 라는 의미이다.
a = set()
a = set([1, 2, 3, 4, 5])
print(a)

# set에 넣으면 list여도 중복을 배제시켜준다.
a = set([1, 2, 3, 4, 5, 4, 2, 1])
print(a)

# set은 수학적으로 집합을 의미하며 집합 연산을 제공한다 : 합집합, 교집합, 차집합
s1 = set([1, 2, 3, 4, 5])
s2 = set([3, 4, 5, 6, 7])

# 합집합 두 가지 방법
print(s1 | s2)      # 연산자 이용
print(s1.union(s2)) # 함수 이용

# 차집합 두 가지 방법
print(s1 - s2)            # 연산자 이용
print(s1.difference(s2))  # 함수 이용

# 교집합 두 가지 방법
print(s1 & s2)              # 연산자 이용
print(s1.intersection(s2))  # 함수 이용

# set은 순서나 key를 가지고 있지 않아 indexing을 할 수 없다
a = set([1, 2, 3, 4, 5])

# set에 값 추가
a.add(10)
print(a)

# set에 값 삭제
a.remove(3)
print(a)
```

```python
# 반복할 횟수가 정해져 있을 때 - for
# 일반적으로 2가지 형태로 사용
# for ~ in range() -> 반복 횟수를 고정해서 사용할 때 10번, 100번
# for ~ in list, dict -> 집합형 자료구조 안에 있는 요소수 만큼 반복

# 1부터 100까지 숫자의 합을 구하기
my_sum = 0 # 값을 누적할 변수
my_range = range(1,101,1) # 1부터 100까지 1씩 증가하는 값들의 모임 *range(초기숫자(포함), 마지막숫자(미포함), 증가값)*

for tmp in my_range : # 100번 반복
  my_sum += tmp
print(my_sum)

a = [1,2,3,4,5]

for tmp in a :
  print(tmp, end = ' ')
```

```python
# Comprehension 개념 (파이썬 기법)
# list comprehension

my_list = [1, 2, 3, 4, 5, 6, 7, 8, 9]
# score1 = my_list * 2  # [1, 2, 3, 4, 5, 6, 7, 8, 9, 1, 2, 3, 4, 5, 6, 7, 8, 9]

score1 = [tmp * 2 for tmp in my_list] # [2, 4, 6, 8, 10, 12, 14, 16, 18]
print(score1)

score2 = [tmp for tmp in my_list if tmp % 2 == 0] # [2, 4, 6, 8]
print(score2)

# 1부터 10까지 각 수의 제곱의 값을 가지고 있는 list를 생성
print('예제문제 1')
my_range = range(1,11) # 1씩 증가는 생략가능하다.
my_score = [tmp**2 for tmp in my_range] # tmp**2 2승 tmp**3 3승 tmp**4 4승
print(my_score)

# list의 각 단어에서 'a'가 포함된 단어들만 필터링
print('예제문제 2')
my_words = ['apple', 'banana', 'cherry','mango']
words_a = [tmp for tmp in my_words if 'a' in tmp]
print(words_a)

# 1부터 10까지 숫자 중 짝수와 그 제곱값을 tuple로 만들어서
# 해당 tuple을 요소로 가지는 새로운 list로 생성
print('예제문제 3') # 잘 모르겠음..
my_list = [(tmp, tmp **2) for tmp in range(1,11) if tmp % 2 == 0] # 만들고자 하는 것을
print(my_list) # [(2, 4), (4, 16), (6, 36), (8, 64), (10, 100)]

#1부터 10까지 숫자 중 list 와 같은 수를 생성하고 싶은데...
my_list = [1,2,3]
list_1 = [ tmp for tmp in range(1,11)]
print(list_1)
```

```python
# Function(함수)

# 반복되고 자주 사용되는 코드를 일련의 코드 단위로 만들어 놓은 것.
# 코드의 재활용성, 관리, 유지보수를 위해 사용함

# 내장함수, 사용자정의 함수가 있다.
# 1. 내장함수 ; type() , 주요한 내장함수는 알고 있는게 좋아요
# 2. 사용자 정의 함수는 우리가 직접 함수를 만들어 사용하는 걸 말한다.
#    사용자 정의 함수의 이름은 특별한 형식으로 만든다. - 관용적으로

# 프로그래밍 언어에서 이름을 짓는 방법. (예를 들어서 변수이름, 함수이름)
# 1. 의미가 있는 이름을 사용합니다.
# 2. 단일 단어는 사용하지 않는다.( 두 단어 이상으로 조합 )
# 3. 이름 짓는 방법
#   - studentName (카멜케이스 노테이션 : 첫글자 소문자, 연결된 단어의 첫글자 대문자)
#   - StudentName (파스칼케이스 노테이션 : 첫글자 대문자, 연결된 단어의 첫글자 대문자)
#   - student_name(스네이크케이스 노테이션 : 두 단어를 _를 이용해서 소문자로 연결) >> 파이썬의 변수와 함수.
```

```python
# 사용자 정의 함수를 만드는 방법
#사용하는 keyword는 def : 정의라는 키워드를 사용한다.

def my_sum(a, b, c) : # a, b, c => 형식인자
  # 함수 호출 했을 때 실행되는 코드들
  #...
  # 거의 예외없이 맨 마지막에 return 구문이 있다.
  return a + b + c

result = my_sum(10, 20, 30) # 10,20,30 => 실인자
print(f'result의 값은 : {result}')
# 기본적인 함수 사용법이다.
```

```python
# 함수로 인자가 전달되는데 ( 실인자 )
# 위에 코드와는 다르게 인자의 수가 가변적이 수 있다.

def my_sum(*args) : # 인자로 전달받은 데이터를 tuple로 받는다.
  result = 0
  for i in range(len(args)) :
    result += args[i]
  return result

print(my_sum(10))
print(my_sum(10,20,30,40))
print(my_sum(10,20))
```

```python
# 많이 혼동(착각)하는것
# 함수의 결과값(리턴값)은 무조건 하나 => 기본 전제
def my_sum(a, b) :
  print(a)
  print(b)
  result1 = a + b
  result2 = a * b
  return result1, result2 # 마치 두 개의 값을 return하는 것처럼 보이지만 tuple 하나를 return 하는 것이다 >> ()생략된 것

output1, output2 = my_sum(10,20)
print(output1, output2)

output3 = my_sum(b = 10, a = 20)  # keyword parameter
```

```python
# Immutable Object , Mutable Object사용이 다르다.
# Immutable Object : 불변객체
#   - int, float, tuple ,range, str, bool
# Mutable Object : 가변 객체
#   - list, dict, set

# 불변 객채 ( Immutable Object )
# 함수에 값을 전달했다 -> 값을 전달 할 때 내가 전달 하려는 값을 복사해서 함수에 전달하는 경우 (별도 값)
# 함수 내부에서 내가 전달한 값을 변경해도 원래 나의 값은 변경되지 않는다.
def modify_value(x) :
  x = x + 10
  print(f'함수 내부에서의 값은? : {x}')

num = 5 # int class (numeric)
modify_value(num)
print(f'함수 외부에서의 값은? : {num}')

# 가변 객체 ( Mutable Object )
def modify_list(lst) : # lst로 list가 전달 된다.
  lst.append(4)
  print(f'함수 내부에서의 list : {lst}')

my_list = [1, 2, 3]
print(f'처음 list : {my_list}')
modify_list(my_list)
print(f'함수 외부에서의 list : {my_list}')
```

```python
# 문제 1 리스트의 합 구하기
# 주어진 리스트의 모든 요소의 합을 구하는 함수를 작성하세요
print('문제 1 리스트 합 구하기')

def sum_of_list(lst) :
  sum_list = 0 # 누적 변수
  for tmp in lst :
    sum_list += tmp
  return sum_list

print(sum_of_list([1, 2, 3, 4]))  # 출력 : 10


# 문제 2 리스트에서 최댓값과 최솟값 구하기
# 리스트에서 최댓값과 최솟값을 구하는 함수를 작성하세요
print('문제 2 최댓값과 최솟값 구하기')
def max_and_min(lst):
  max = 0
  min = 0
  for tmp in lst :
    if max < tmp :
      max = tmp
    elif max > tmp :
      min = tmp
  return max, min

print(max_and_min([10,20,4,45,99])) # 출력 : (99,4)
```
