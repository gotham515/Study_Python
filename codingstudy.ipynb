```python
# 문제 1: 리스트의 합 구하기
# 주어진 리스트의 모든 요소의 합을 구하는 함수를 작성하세요.

def sum_of_list(lst):
    result = 0
    for tmp in lst :
        result += tmp
    return result

print(sum_of_list([1, 2, 3, 4]))  # 출력: 10
```

```python
# 문제 2: 리스트에서 최댓값과 최솟값 구하기
# 리스트에서 최댓값과 최솟값을 구하는 함수를 작성하세요.

def max_and_min(lst):
    max, min = 0, 0
    for tmp in lst :
        if max < tmp :
            max = tmp
        elif max > tmp :
            min = tmp
    return max, min

print(max_and_min([10, 20, 4, 45, 99]))  # 출력: (99, 4)
```

```python
# 문제 3: 리스트 뒤집기
# 주어진 리스트를 역순으로 뒤집는 함수를 작성하세요.
# (직접 구현해 보세요, list.reverse() 사용하지 말 것)

def reverse_list(lst):
    rev_list = []
    for idx in range(len(lst)-1, -1, -1) :
        rev_list.append(lst[idx])
    return rev_list

print(reverse_list([1, 2, 3, 4]))  # 출력: [4, 3, 2, 1]
```

```python
# 문제 4: 리스트 중복 제거
# 리스트에서 중복된 요소를 제거하고, 중복이 제거된 리스트를 반환하는 함수를 작성하세요.

def remove_duplicates(lst):
    set_list = set(lst)
    return list(set_list)

print(remove_duplicates([1, 2, 2, 3, 4, 4, 5]))  # 출력: [1, 2, 3, 4, 5]
```

```python
#문제 5: 리스트 요소들의 곱 구하기
# 리스트의 모든 요소의 곱을 구하는 함수를 작성하세요.

def multiply_list(lst):
    result = 1
    for tmp in lst :
        result *= tmp
    return result

print(multiply_list([1, 2, 3, 4]))  # 출력: 24
```

```python
# 문제 6: 리스트에서 짝수와 홀수 구분하기
# 주어진 리스트에서 짝수와 홀수를 구분하여 별도의 리스트에 저장한 뒤 반환하는 함수를 작성하세요.

def separate_even_odd(lst):
    lst_even = []
    lst_odd = []

    for tmp in lst :
        if tmp % 2 == 0 :
            lst_odd.append(tmp)
        else :
            lst_even.append(tmp)

    return lst_odd, lst_even

print(separate_even_odd([1, 2, 3, 4, 5, 6, 7, 8, 9]))  # 출력: ([2, 4, 6, 8], [1, 3, 5, 7, 9])
```

```python
# 문제 7: 리스트 요소들의 위치 교환
# 리스트의 첫 번째 요소와 마지막 요소의 위치를 바꾸는 함수를 작성하세요.

def swap_first_last(lst):
    tmp_index = lst[-1]
    lst[-1] = lst[0]
    lst[0] = tmp_index
    return lst

print(swap_first_last([1, 2, 3, 4, 5]))  # 출력: [5, 2, 3, 4, 1]
```

```python
# 문제 8: 리스트에서 특정 요소의 모든 인덱스 찾기
# 리스트에서 특정 요소가 나타나는 모든 인덱스를 리스트로 반환하는 함수를 작성하세요.

def find_all_indices(lst, value):
    list_enu = enumerate(lst)
    result_list = [idx for idx, tmp in list_enu if value == tmp]
    return result_list

print(find_all_indices([1, 2, 3, 2, 4, 2, 5], 2))  # 출력: [1, 3, 5]
```

```python
# 문제 9: 리스트의 중간 요소 찾기
# 리스트의 중간에 있는 요소를 반환하는 함수를 작성하세요.
# 리스트의 길이가 짝수일 경우 두 개의 중간 요소를 반환하세요.

def find_middle(lst):
    find_num = ()
    if len(lst) % 2 == 0 :
        find_num = lst[len(lst)//2-1], lst[len(lst)//2]
    else :
        find_num = lst[len(lst)//2]
    return find_num

print(find_middle([1, 2, 3, 4, 5]))  # 출력: 3
print(find_middle([1, 2, 3, 4]))     # 출력: (2, 3)
print(find_middle([1, 2, 3, 4, 5, 6]))     # 출력: (3, 4)
```

```python
# 문제 10: 중첩 리스트 펼치기 (Flattening)
# 중첩된 리스트를 평평하게 만들어 단일 리스트로 반환하는 함수를 작성하세요.

def flatten(lst):
    list_enu = enumerate(lst)
    new_list = []
    for idx,tmp in list_enu :
        if type(tmp) == list :
            ls = enumerate(tmp)
            for idx, tmp2 in ls :
                if type(tmp2) == list:
                    new_list += tmp2
                elif type(tmp2) == int :
                    new_list.append(tmp2)
        elif type(tmp) == int :
            new_list.append(tmp)
    return new_list

print(flatten([1, [2, 3], [4, [5, 6]], 7]))  # 출력: [1, 2, 3, 4, 5, 6, 7]
```

```python
# 문제 1: 딕셔너리에서 값의 합 구하기
# 주어진 딕셔너리에서 모든 값의 합을 구하는 함수를 작성하세요.
# (딕셔너리의 값은 모두 숫자라고 가정합니다.)

def sum_of_dict_values(d):
    dict_values = d.values()
    result = 0
    for tmp in dict_values :
        result += tmp
    return result

print(sum_of_dict_values({'a': 100, 'b': 200, 'c': 300}))  # 출력: 600
```

```python
# 문제 3: 키 값 뒤집기
# 주어진 딕셔너리에서 키와 값을 뒤집어 새로운 딕셔너리를 반환하는
# 함수를 작성하세요. (값은 고유하다고 가정합니다.)

def invert_dict(d):
    new_dict = {}
    for key in d :
        new_dict_key_value = d[key]
        new_dict[new_dict_key_value] = key
    return new_dict

print(invert_dict({'a': 1, 'b': 2, 'c': 3}))  # 출력: {1: 'a', 2: 'b', 3: 'c'}
```

```python
# 문제 4: 두 딕셔너리 합치기
# 두 개의 딕셔너리를 합쳐서 새로운 딕셔너리를 반환하는 함수를 작성하세요.
# 만약 동일한 키가 존재하면 두 값의 합을 사용하세요.

def merge_dicts(d1, d2):
    new_dict = {}
    for key in d1 :
        new_dict[key] = d1[key]
    for key in d2 :
        new_dict[key] = d2[key]

    for key in d1 :
        for key2 in d2 :
            if key == key2 :
                d1[key] +=  d2[key]
                new_dict[key] = d1[key]
    return new_dict

print(merge_dicts({'a': 100, 'b': 200}, {'b': 300, 'c': 400}))  # 출력: {'a': 100, 'b': 500, 'c': 400}
```

```python
# 문제 6: 중첩 딕셔너리에서 값 가져오기
# 중첩된 딕셔너리에서 특정 키의 값을 가져오는 함수를 작성하세요.
# 키가 없으면 None을 반환합니다.

def get_nested_value(d, keys):
    result = d
    for n in keys :
        if n not in result :
            return None
        result = result[n]
    return result


nested_dict = {'a': {'b': {'c': 100}}}
print(get_nested_value(nested_dict, ['a', 'b', 'c']))  # 출력: 100
print(get_nested_value(nested_dict, ['a', 'b', 'x']))  # 출력: None
```

```python
# 문제 7: 키와 값 스왑하기
# 딕셔너리에서 키와 값을 스왑하고, 동일한 값을 가진 키들이 있을 경우
# 리스트로 묶는 함수를 작성하세요.

def swap_keys_values(d):
    new_dict = {}
    for key in d :
        key_value = d[key]
        if key_value not in new_dict :
            new_dict[key_value] = key
        elif key_value in new_dict :
            new_dict[key_value] = [new_dict[key_value]] + [key]
    return new_dict

print(swap_keys_values({'a': 1, 'b': 2, 'c': 1}))  # 출력: {1: ['a', 'c'], 2: 'b'}
```

```python
# 문제 8: 딕셔너리 필터링
# 주어진 딕셔너리에서 특정 값을 가지는 키-값 쌍만 남겨 새로운 딕셔너리를
# 반환하는 함수를 작성하세요.

def filter_dict_by_value(d, value):
    new_dict = {}
    for key, val in d.items() :
        if val == value :
            new_dict[key] = val

    return new_dict

print(filter_dict_by_value({'a': 100, 'b': 200, 'c': 100}, 100))
# 출력: {'a': 100, 'c': 100}
```

```python
# 문제 9: 두 딕셔너리의 공통 키 찾기
# 두 딕셔너리에서 공통된 key와 value를 가진 모든 키-값 쌍을 반환하는 함수를 작성하세요.

def common_keys(d1, d2):
    new_dict = {}
    for key1, value1 in d1.items() :
        for key2, value2 in d2.items() :
            if key1 == key2 and value1 == value2 :
                new_dict[key1] = d1[key1]
    return new_dict

print(common_keys({'a': 100, 'b': 200, 'c': 300}, {'b': 200, 'c': 400, 'd': 500}))
# 출력: {'b': 200}
```

```python
# 문제 10: 리스트의 딕셔너리로 변환
# 두 개의 리스트를 받아서 하나는 키, 다른 하나는 값으로 사용하여 딕셔너리를
# 생성하는 함수를 작성하세요.
# 리스트의 길이가 다르면 더 짧은 길이에 맞춰서 딕셔너리를 만드세요.

def lists_to_dict(keys, values):
    new_dict = {}
    dict_range = 0

    if len(keys) > len(values) :
        dict_range = len(values)
    elif len(keys) < len(values) :
        dict_range = len(keys)
    else :
        dict_range = len(keys)

    for idx in range(dict_range) :
        key_value = keys[idx]
        if key_value not in new_dict :
            new_dict[key_value] = values[idx]

    return new_dict


keys = ['a', 'b', 'c']
values = [1, 2, 3]
print(lists_to_dict(keys, values))  # 출력: {'a': 1, 'b': 2, 'c': 3}
```

```python
# 문제 1: 숫자 뒤집기
# 주어진 정수의 자리수를 반대로 뒤집는 함수를 작성하세요.
# (음수도 처리해야 합니다)

def reverse_integer(n):
    str_n = str(n)
    new_str = ''

    for tmp in range(len(str_n)) :
        if str_n[0] == '-' :
           new_str = '-' + str_n[-1:0:-1]
        else :
           new_str = str_n[-1:0:-1]

    return int(new_str)

print(reverse_integer(12345))  # 출력: 54321
print(reverse_integer(-12345))  # 출력: -54321
```

```python

# 문제 2: 문자열 문자 수 세기
# 주어진 문자열에서 각 문자의 발생 횟수를 세고, 이를 딕셔너리 형태로 반환하는
# 함수를 작성하세요.

def count_characters(s):
    lst_dict = {}
    for key in s :
        if key not in lst_dict:
            lst_dict[key] = 1
        else:
            lst_dict[key] += 1
    return lst_dict

print(count_characters("hello"))  # 출력: {'h': 1, 'e': 1, 'l': 2, 'o': 1}
```

```python
# 문제 3: 튜플과 리스트 변환
# 튜플을 리스트로, 리스트를 튜플로 변환하는 함수를 작성하세요.

def convert_tuple_list(data):
    if type(data) == tuple :
        return list(data)
    elif type(data) == list :
        return tuple(data)

print(convert_tuple_list((1, 2, 3)))  # 출력: [1, 2, 3]
print(convert_tuple_list([1, 2, 3]))  # 출력: (1, 2, 3)
```

```python
# 문제 4: 집합 교집합, 합집합, 차집합
# 두 개의 리스트를 받아서 각각 교집합, 합집합, 차집합을 구하는 함수를 작성하세요.

def set_operations(list1, list2) :
    lst_set1 = set(list1)
    lst_set2 = set(list2)
    return { 'intersection' : lst_set1.intersection(lst_set2) , 'union' : lst_set1.union(lst_set2) , 'defference' : lst_set1.difference(lst_set2)  }

print(set_operations([1, 2, 3], [3, 4, 5])) # 출력: {'intersection': {3}, 'union': {1, 2, 3, 4, 5}, 'difference': {1, 2}}
```

```python
# 문제 5: 리스트 중간 요소 찾기
# 주어진 리스트의 중간 요소를 반환하는 함수를 작성하세요.
# 만약 리스트의 길이가 짝수이면, 두 개의 중간 요소를 반환하세요.

def find_middle_element(lst):
    find_num = ()
    if len(lst) % 2 == 0 :
        find_num = lst[len(lst)//2-1], lst[len(lst)//2]
    else :
        find_num = lst[len(lst)//2]
    return find_num

print(find_middle_element([1, 2, 3, 4, 5]))  # 출력: 3
print(find_middle_element([1, 2, 3, 4]))     # 출력: (2, 3)
```

```python
# 문제 6: 키-값 쌍 리스트 변환
# 딕셔너리에서 키와 값을 각각 리스트로 변환하는 함수를 작성하세요.

def dict_to_lists(d):
    dict_key = d.keys()
    dict_value = d.values()
    return list(dict_key) ,list(dict_value)

print(dict_to_lists({'a': 1, 'b': 2, 'c': 3}))
# 출력: (['a', 'b', 'c'], [1, 2, 3])
```

```python
# 문제 7: 문자열에서 단어의 역순 출력
# 주어진 문자열에서 단어의 순서를 뒤집어 출력하는 함수를 작성하세요.

def reverse_words(s):
    first_str = ''
    last_str = ''
    space_str = 0

    for tmp in s :
        space_str += 1
        if tmp == ' ':
            first_str = s[0 : space_str -1]
            last_str = s[space_str-1 :]

    return last_str +' '+ first_str

print(reverse_words("Hello World"))  # 출력: "World Hello"
```

```python
# 문제 8: 중복 요소 제거
# 주어진 리스트에서 중복된 요소를 제거한 리스트를 반환하는 함수를 작성하세요.

def remove_duplicates(lst):
    lst_set = set(lst)
    return list(lst_set)

print(remove_duplicates([1, 2, 2, 3, 4, 4, 5]))  # 출력: [1, 2, 3, 4, 5]
```

```python
# 문제 9: 모든 요소 곱하기
# 리스트 안의 모든 요소를 곱한 값을 반환하는 함수를 작성하세요.

def multiply_elements(lst):
    result = 1;
    for value in lst :
        result *= value
    return result

print(multiply_elements([1, 2, 3, 4]))  # 출력: 24
```

```python
# 문제 10: 짝수 인덱스 요소 추출
# 주어진 리스트에서 짝수 인덱스에 위치한 요소들만 추출하여
# 새로운 리스트를 반환하는 함수를 작성하세요.

def even_index_elements(lst):
    lst_enu = enumerate(lst)
    new_lst = []
    for idx, value in lst_enu :
        if idx % 2 == 0 :
            new_lst.append(value)
    return new_lst

print(even_index_elements([10, 20, 30, 40, 50]))  # 출력: [10, 30, 50]
```

```python
# 문제 11: 문자열을 정수로 변환
# 문자열로 표현된 숫자를 정수로 변환하는 함수를 작성하세요.
# 단, 숫자 이외의 문자가 포함된 경우에는 None을 반환하세요.

def str_to_int(s):
    words_a = [tmp for tmp in s if 'a' in tmp]

    for tmp in words_a :
        for tmp2 in s :
            if tmp == tmp2 :
                return None
    return s


print(str_to_int("12345"))  # 출력: 12345
print(str_to_int("12a45"))  # 출력: None
```

```python
# 문제 13: 리스트 합치기
# 두 개의 리스트를 하나로 합치는 함수를 작성하세요.

def concatenate_lists(lst1, lst2):
    return lst1 + lst2

print(concatenate_lists([1, 2, 3], [4, 5, 6]))  # 출력: [1, 2, 3, 4, 5, 6]
```

```python
# 문제 14: 문자열에서 모음 개수 세기
# 주어진 문자열에서 모음(a, e, i, o, u)의 개수를 세는 함수를 작성하세요.

def count_vowels(s):
    vow_cnt = ['a','e','i','o','u']
    cnt = 0
    for tmp in s :
        for tmp2 in vow_cnt:
            if tmp == tmp2 :
                cnt += 1
    return cnt

print(count_vowels("hello world"))  # 출력: 3
```

```python
# 문제 15: 중첩 리스트 평탄화
# 중첩된 리스트를 평탄화하여 단일 리스트로 만드는 함수를 작성하세요.

def flatten_list(lst):
    list_enu = enumerate(lst)
    new_list = []
    for idx,tmp in list_enu :
        if type(tmp) == list :
            ls = enumerate(tmp)
            for idx, tmp2 in ls :
                if type(tmp2) == list:
                    new_list += tmp2
                elif type(tmp2) == int :
                    new_list.append(tmp2)
        elif type(tmp) == int :
            new_list.append(tmp)
    return new_list

print(flatten_list([1, [2, 3], [4, [5, 6]], 7]))
# 출력: [1, 2, 3, 4, 5, 6, 7]
```
